```python
import os

# Đường dẫn tới thư mục cha
parent_dir = "/ubuntu/1/BTL_CSDLDPT/selected_test_folders"

# Duyệt qua từng folder con
subfolders = [f for f in os.listdir(parent_dir) if os.path.isdir(os.path.join(parent_dir, f))]

for folder in subfolders:
    folder_path = os.path.join(parent_dir, folder)
    # Lấy danh sách các file ảnh (đuôi jpg, jpeg, png)
    images = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    # Giữ lại 10 ảnh đầu tiên
    images_to_keep = images[:10]
    # Xóa các ảnh còn lại
    images_to_delete = images[10:]
    for image in images_to_delete:
        image_path = os.path.join(folder_path, image)
        os.remove(image_path)
        print(f"Đã xóa: {image_path}")

print("Hoàn thành việc giữ lại 10 ảnh đầu trong mỗi folder con.")
```

```python
import os
import shutil

# Đường dẫn tới thư mục cha
parent_dir = "/home/ubuntu/1/BTL_CSDLDPT/selected_test_folders"
output_dir = "/home/ubuntu/1/BTL_CSDLDPT/DataLoc"

# Tạo thư mục DataLoc nếu chưa tồn tại
os.makedirs(output_dir, exist_ok=True)

# Duyệt qua từng folder con
subfolders = [f for f in os.listdir(parent_dir) if os.path.isdir(os.path.join(parent_dir, f))]

for folder in subfolders:
    folder_path = os.path.join(parent_dir, folder)
    output_folder_path = os.path.join(output_dir, folder)
    os.makedirs(output_folder_path, exist_ok=True)  # Tạo folder con trong DataLoc

    # Lấy danh sách các file ảnh (đuôi jpg, jpeg, png)
    images = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))][:10]

    # Sao chép 10 ảnh đầu tiên vào folder mới
    for image in images:
        src_path = os.path.join(folder_path, image)
        dest_path = os.path.join(output_folder_path, image)
        shutil.copy(src_path, dest_path)
        print(f"Đã sao chép: {src_path} -> {dest_path}")

print("Hoàn thành việc tạo thư mục DataLoc với mỗi folder con chứa tối đa 10 ảnh.")
```

```python
import os
import pymongo
import gridfs

# Kết nối tới MongoDB
client = pymongo.MongoClient("mongodb://localhost:27017/")
db = client["image_database"]  # Tên cơ sở dữ liệu
fs = gridfs.GridFS(db)  # Sử dụng GridFS để lưu ảnh nhị phân
collection = db["image_data"]  # Tên collection để lưu thông tin ảnh

# Đường dẫn tới thư mục cha
parent_dir = "/home/ubuntu/1/BTL_CSDLDPT/selected_test_folders"

# Duyệt qua từng folder con
subfolders = [f for f in os.listdir(parent_dir) if os.path.isdir(os.path.join(parent_dir, f))]

for folder in subfolders:
    folder_path = os.path.join(parent_dir, folder)
    images = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

    for image_name in images:
        image_path = os.path.join(folder_path, image_name)

        # Lưu ảnh vào GridFS
        with open(image_path, "rb") as img_file:
            image_id = fs.put(img_file, filename=image_name)

        # Tạo tài liệu MongoDB
        document = {
            "folder_name": folder,
            "image_name": image_name,
            "image_id": image_id  # ID của ảnh trong GridFS
        }

        # Chèn tài liệu vào MongoDB
        collection.insert_one(document)
        print(f"Đã lưu {image_name} từ folder {folder} vào MongoDB.")

print("Hoàn thành lưu trữ ảnh vào MongoDB.")
```
